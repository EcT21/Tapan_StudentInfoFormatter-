```python
#Version 2

from datetime import datetime

event = input("Enter event name: ")
year = int(input("Enter year: "))
month = int(input("Enter month (numerical): "))
day = int(input("Enter day: "))

event_info = [event, year, month, day]
event_date = datetime  (event_info[1], event_info[2], event_info[3])

month_name = event_date.strftime("%B")
weekday = event_date.strftime("%A")

print("\nFormatted Output: ")
print("Event: ", event)
print("Date", f"{month}/{day}/{year}")
print("Year: ", year)
print("Month: ", month)
print("Day: ", day)

def info_display():
  for i in range(len(event_info)):
    if i == 0:
      print("Event: ", event_info[i].upper())
    if i == 1:
      print("Date: ", month_name, event_info[3], ",", event_info[1])
      print("Year: ", event_info[i])
    if i == 2:
      print("Month: ", month_name)
    if i == 3:
      print("Day: ", event_info[i])
      print("Weekday: ", weekday)

print("\n Formatted Output")
print("\n Event Information")
info_display()

#Version 1 only displays basic date information, while Verion 2 uses datetime, lists, loops, functions, and conditional statements to display more detailed date information
```
