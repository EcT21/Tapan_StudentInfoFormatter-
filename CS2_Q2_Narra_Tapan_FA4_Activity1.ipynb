```python
#Version 1

first = str(input("Enter first name: "))
last = str(input("Enter last name: "))
section = str(input("Enter section: "))
favSubject = str(input("Enter favorite subject: "))

name = [first, last]
all = [first, last, section, favSubject]
count = len(" ".join(all))

print("\nFormatted Output: ")
print("Name : ", " ".join(name))
print("NAME CODE:", first.upper(), last.upper())
print("Section: ", section.upper())
print("Favorite Subject: ", favSubject)
print("Character Count", (count))
```
